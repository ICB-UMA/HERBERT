```python
import os
import pandas as pd
```

```python
OUTPUT_PATH = "../output/"
```

```python
MODEL_NAME_MAP = {
    "SapBERT-UMLS-2020AB-all-lang-from-XLMR-large": "SapBERT-XLMR-large",
    "SapBERT-UMLS-2020AB-all-lang-from-XLMR": "SapBERT-XLMR-base",
}

RECALL_COLS = ["Recall@1", "Recall@5", "Recall@25", "Recall@200"]
RENAME_RECALL = {
    "Recall@1": "R@1",
    "Recall@5": "R@5",
    "Recall@25": "R@25",
    "Recall@200": "R@200",
}
```

```python
def _load_and_format(path: str, split_name: str) -> pd.DataFrame:
    df = pd.read_csv(path, sep="\t")
    if "name" not in df.columns:
        raise ValueError(f"Missing 'name' column in {path}")

    for c in RECALL_COLS:
        if c not in df.columns:
            df[c] = pd.NA

    df = df[["name"] + RECALL_COLS].copy()

    df["name"] = df["name"].map(lambda x: MODEL_NAME_MAP.get(x, x))
    df = df.set_index("name")
    df = df.rename(columns=RENAME_RECALL)
    df = df.astype(float).round(3)

    df.columns = pd.MultiIndex.from_product([[split_name], df.columns.tolist()])
    return df
```

### DisTEMIST

```python
corpus = "DisTEMIST"
aux_path = os.path.join(OUTPUT_PATH, corpus)

gs_path = os.path.join(aux_path, "results_gs.tsv")
um_path = os.path.join(aux_path, "results_um.tsv")
uc_path = os.path.join(aux_path, "results_uc.tsv")
```

```python
df_gs = _load_and_format(gs_path, "Gold Standard")
df_um = _load_and_format(um_path, "Unseen Mentions")
df_uc = _load_and_format(uc_path, "Unseen Codes")

df_all = pd.concat([df_gs, df_um, df_uc], axis=1)

split_order = ["Gold Standard", "Unseen Mentions", "Unseen Codes"]
metric_order = ["R@1", "R@5", "R@25", "R@200"]
df_all = df_all.reindex(
    columns=pd.MultiIndex.from_product([split_order, metric_order])
)

df_all
```
